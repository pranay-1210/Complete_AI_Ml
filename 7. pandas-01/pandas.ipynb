```python
# Series in Pandas - 1D labelled Array

s = pd.Series([10,20,30,40,50])
print(s)
print(type(s))

#index:-

print(s[3])
print(s[2])
```

```python
# Custom Labels:-

s = pd.Series([23,28,73,29], index=["Adam","Eve","Bob","Charlie"])
print(s)

#Accessing on the basis of our custom labels:-

print(s["Bob"])
print(s["Charlie"])
print(s.index)

```

```python
# Properties of Series:-

# vectorized operations:

s1 = pd.Series([1,2,3,4,5,6])
s2 = pd.Series([7,8,9,10,11,12])

print(s1 + s2)


```

```python
# Mutable
s2 = pd.Series([1,2,3,4,5,6])
s2[0] = 100
print(s2)
```

```python
# Filtering of Data:-

df[df["aqi"] > 100]
```

```python
#Multiple condition based filtering:-

df[(df["aqi"] > 100) & (df["temperature"] < 18)]
```

```python
# Filtering of Data with the help of query method:-
# Query Method by default returns a copy of the filtered data not a view.

#RULES:-

## Conditions are passed as a string.
## Can access using the column name.
## Can also use backticks instead of string.
## Can also use operators inside the string.
## Can also use python variables under the query method using the @ as reference.

df.query("aqi > 100")

aqi_val = 100
df.query("aqi > @aqi_val")

```
