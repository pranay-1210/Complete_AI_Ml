Write a web scraping program to collect all quotes and corresponding authors from all pages that include the life tag. Ensure pagination is handled properly.

```python
import requests
from bs4 import BeautifulSoup
```

```python
#Fetching and Storing the Data:

page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text,"lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore..")
        break

    with open(f"scrapped_data2/quotes{page_count}.html","w",encoding="utf-8") as f:
        f.write(res.text)
        print(f"downloaded data from page{page_count}")

    page_count = page_count + 1
    
    
    
```

```python
from bs4 import BeautifulSoup

life_quotes = []

for page in range(1, 11):
    with open(f"scrapped_data2/quotes{page}.html", "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")

    for q in all_quotes:
        all_tags = [tag.get_text() for tag in q.select(".tags .tag")]

        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append((text, author))

print(len(life_quotes))
print(life_quotes)
```

```python
#Storing collected Data:

import pandas as pd

df = pd.DataFrame(life_quotes,columns=["Quote","Author"])
df
```

```python
df.to_csv("cleaned_data2/data2.csv",index=False)
```
